```python
import numpy as np
from tensorflow.keras.datasets import fashion_mnist
from tensorflow.keras.utils import to_categorical

# 데이터 로드
(x_train, y_train), (x_test, y_test) = fashion_mnist.load_data()

# 정규화
x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0

# Flatten
x_train = x_train.reshape(-1, 784)
x_test = x_test.reshape(-1, 784)

# One-hot encoding
y_train_cat = to_categorical(y_train, 10)
y_test_cat = to_categorical(y_test, 10)

print(x_train.shape)
print(x_test.shape)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.layers import BatchNormalization
from tensorflow.keras.layers import Dropout

model = Sequential([
    Dense(512, input_shape=(784,)),
    BatchNormalization(),
    Dense(512, activation='relu'),
    Dropout(0.3),

    Dense(256),
    BatchNormalization(),
    Dense(256, activation='relu'),
    Dropout(0.3),

    Dense(128, activation='relu'),

    Dense(10, activation='softmax')
])

model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

model.summary()
```

```python
history = model.fit(
    x_train,
    y_train_cat,
    validation_split=0.2,
    epochs=20,
    batch_size=128,
    verbose=1
)
```

```python
test_loss, test_acc = model.evaluate(
    x_test,
    y_test_cat,
    verbose=0
)

print("Test Accuracy:", test_acc)
```

```python
from sklearn.metrics import confusion_matrix
from sklearn.metrics import classification_report

pred = model.predict(x_test)
pred_class = np.argmax(pred, axis=1)

cm = confusion_matrix(y_test, pred_class)

print(classification_report(y_test, pred_class))
```

- 결과 해석: 모델이 정확도가 0.879정도로 양호한 편이다.
- 학습을 할수록 Accuracy가 증가하였고, Loss는 감소하였다. 모델이 학습을 잘 한것으로 보인다.
- Epoch에서 Training Accuracy는 계속 증가하였으나 Validation Loss도 살짝 증가하는걸로 보아 약한 과적합이 있는걸로 보인다.
- 한계: DNN은 이미지를 1차원 벡터로 변환하여 처리하기 때문에 공간 정보를 충분히 활용하지 못하는 한계가 있다.
